```python
import os
import re

def set_supervision(value):
    assert value in ("semi", "weak")
    for folder in ("cartpole", "lunarlander"):
        for fn in ("vae_p3.py", "lstm_p3.py"):
            fp = os.path.join(folder, fn)
            if not os.path.exists(fp):
                continue
            with open(fp, "r") as f:
                content = f.read()
            new = re.sub(r'SUPERVISION = "[^"]*"', f'SUPERVISION = "{value}"', content, count=1)
            if new != content:
                with open(fp, "w") as f:
                    f.write(new)

# !python3 cartpole/dataCollect.py
# !python3 lunarlander/dataCollect.py
```
